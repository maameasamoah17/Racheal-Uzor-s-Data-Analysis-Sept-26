### Task 1: Basic if Statement

```python
# Check if a number is positive
def check_positive(number):
    if number > 0:
        print("The number is positive")
# Example
check_positive(5)
```

### Task 2: if-else statement 

```python
# Determine if a student passed or failed
def check_result(score):
    if score >= 60:
        print("Passed")
    else:
        print("Failed")
# Example
check_result(75)
check_result(45)
```

### Task 3: if-elif-else statement

```python
# Convert a numeric grade to a letter grade
def get_letter_grade(grade):
    if grade >= 90 and grade <= 100:
        print("A")
    elif grade >= 80:
        print("B")
    elif grade >= 70:
        print("C")
    elif grade >= 60:
        print("D")
    elif grade >= 0:
        print("F")
    else:
        print("Invalid grade")
# Example
get_letter_grade(93)
get_letter_grade(85)
get_letter_grade(54)
```

### Task 4: Nested Conditionals

```python
# Determine access based on login and membership status
def check_access(is_logged_in, is_admin, is_premium):
    if not is_logged_in:
        print("Please log in to access the site")
    elif is_admin:
        print("Welcome admin! Full access granted")
    elif is_premium:
        print("Welcome premium user! Premium content access granted")
    else:
        print("Welcome! Basic access granted")

# Example 
check_access(False, False, False)
check_access(True, True, False)
check_access(True, False, True)
check_access(True, False, False)
```

### Task 5: Ternary Operator

```python
# Return whether a number is even or odd
def check_even_odd(number):
    return "Even" if number % 2 == 0 else "Odd"

# Example 
print(check_even_odd(8))
print(check_even_odd(7))
```

### Task 6: Combining Conditions

```python
# Determine whether a year is a leap year
def check_leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        print("Leap year")
    else:
        print("Not a leap year")

# Example 
check_leap_year(2024)
check_leap_year(1900)
check_leap_year(2000)
```

### Task 7: Best Practices (Refactoring)

```python
# Classify an item based on type, price, and availability
def classify_item(item_type, price, in_stock):
    # Determine the type of item
    if item_type == "electronics":
        category = "electronics"
    else:
        category = "non-electronics"

    # Determine whether the item is affordable or premium
    if price < 100:
        price_category = "Affordable"
    else:
        price_category = "Premium"

    # Determine the stock status
    if in_stock:
        stock_status = "in stock"
    else:
        stock_status = "out of stock"

    # Combine the results into one message
    return f"{price_category} {category} {stock_status}"

# Example 
print(classify_item("electronics", 75, True))
print(classify_item("electronics", 150, False))
print(classify_item("clothing", 50, True))
```
