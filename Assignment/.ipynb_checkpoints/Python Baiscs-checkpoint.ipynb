### Task 1: Variable Declaration and Data Types  

```python
# Declare variables of different types (integer, float, string, boolean, etc.).
age = 67 #integer 
hourly_rate = 31.26 #float
course_name = "Data Analysis" #String
is_enrolled = True #Boolean

# Print their values and types using the type() function.
print(age, type(age))
print(hourly_rate, type(hourly_rate))
print(course_name, type(course_name))
print(is_enrolled, type(is_enrolled))
```

### Task 2: String Manipulation

```python
# Create a string variable.
snack_message = "I want some Cheetos!"

# Apply at least three different string methods to modify or retrieve information from the string.

print(snack_message.upper()) # Convert to uppercase.

print(snack_message.replace("Cheetos", "Apple")) # Replaces Cheetos with Apple.

print(snack_message.title()) # Capitalizes the first letter of each word.
```

### Task 3: Basic Operations

```python
# Perform and print the results of the following operations
x = 17
y = 9

# Arithmetic
print(x + y) # Addition
print(x - y) # Subtraction
print(x ** y) # Exponentiation
print(x * y) # Multiplication
print(x / y) # Division
print(x % y) # Remainder (modulus)


# Comparison Operations
print(x > y)   # Checks if x is greater than y
print(x < y)   # Checks if x is less than y
print(x == y)  # Checks if x is equal to y
print(x != y)  # Checks if x is not equal to y
print(x >= y)  # Checks if x is greater than or equal to y
print(x <= y)  # Checks if x is less than or equal to y

# Logical Operations
print(x > 16 and y < 4)  # True only if both conditions are True
print(x < 10 or y > 6)   # True if at least one condition is True
print(not (x == y))      # Reverses the result: True becomes False, and False becomes True
```

### Task 4: Data Collections

```python
# Create a list of numbers
scores = [85, 90, 76, 95, 88]

# Create a dictionary with student information
student_info = {"name": "Racheal", "age": 67}

# Create a tuple of numbers
test_scores = (70, 82, 91)

# Access an element from the list using its index
print(scores[2])  # Prints the third element: 76

# Access a value from the dictionary using its key
print(student_info["name"])  # Prints the name: Racheal

# Access an element from the tuple using its index
print(test_scores[1])  # Prints the second element: 82
```

### Task 5: Expression Combination

```python
price_per_item = 12
quantity = 4
discount = 5
tax = 3

# Calculate the total cost after subtracting the discount and adding tax
total_cost = (price_per_item * quantity) - discount + tax

# Print the final result
print(total_cost)
```
