```python
# Calculate a number raised to a power
def power(base, exponent=2):
    return base ** exponent

# Example
print(power(4))     # Uses the default exponent of 2
print(power(4, 3))  # Uses an exponent of 3
```

### Task 5: Function that Returns a Value

```python
# Combine first, middle, and last names
def get_full_name(first_name, last_name, middle_name=""):
    if middle_name:
        return f"{first_name} {middle_name} {last_name}"
    else:
        return f"{first_name} {last_name}"

# Example
print(get_full_name("Alice", "Smith"))
print(get_full_name("Alice", "Smith", "Marie"))
```

### Task 7: Combining Functions and Conditionals

```python
# Convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

# Convert Fahrenheit to Celsius
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

# Analyze temperature and describe the weather
def analyze_temperature(temperature, unit):
    if unit == "C":
        celsius = temperature
    elif unit == "F":
        celsius = fahrenheit_to_celsius(temperature)
    else:
        return "Invalid unit. Use 'C' or 'F'."

    if celsius < 0:
        description = "freezing"
    elif celsius < 15:
        description = "cold"
    elif celsius <= 25:
        description = "moderate"
    else:
        description = "hot"

    return description

# Example
print(analyze_temperature(5, "C"))
print(analyze_temperature(77, "F"))
```

### Task 8: Function with Variable Arguments

```python
# Calculate the average of any number of values
def calculate_average(*numbers):
    if len(numbers) == 0:
        return 0

    return sum(numbers) / len(numbers)

# Example 
print(calculate_average(10, 20, 30))
print(calculate_average(75, 85, 95, 100))
```

### Task 9: Advanced Challenge — Student Grade Function

```python
# Calculate a student's average and letter grade
def calculate_final_grade(name, *scores):
    if len(scores) == 0:
        return f"No scores provided for {name}."

    average = sum(scores) / len(scores)

    if average >= 90:
        letter_grade = "A"
    elif average >= 80:
        letter_grade = "B"
    elif average >= 70:
        letter_grade = "C"
    elif average >= 60:
        letter_grade = "D"
    else:
        letter_grade = "F"

    return f"{name}'s final grade is {average:.0f} ({letter_grade})."

# Example 
print(calculate_final_grade("Alice", 80, 85, 90))
```
