### Task 2: String Manipulation

```python
sample_string = "Hello, Python!"
# Convert to lowercase
print(sample_string.lower()) 

# Replace part of the string 
print(sample_string.replace("Hello", "Hi"))

#Split into a list
print(sample_string.split(","))
```

### Task 3: Basic Operations

```python
a,b = 17, 8

# Arithmetic
print(a + b, a - b, a * b, a / b, a % b, a ** b)

# Comparison
print(a > b, a < b, a == b, a != b, a >= b, a <= b)

# Logical
print(a > 16 and b < 4, a < 10 or b > 6, not (a == b))
```

### Task 4: Data Collections

```python
my_list = [1, 2, 3, 4, 5]
my_dict = {"name": "Leticia", "age": 27}
my_tuple = (10, 20, 30)

# Access element from list
print(my_list[3])

# Access element from dictionary
print(my_dict["name"])

# Access element from tuple
print(my_tuple[2])
```
